**Q1) Write a Python function to find the maximum and minimum numbers from a
sequence of numbers.  (Note: Do not use built-in functions.) .**

```python
def max_min(l):
    max = l[0]
    min = l[0]

    for i in l:
        if i > max:
            max = i

        if i < max:
            min = i

    return max,min

l=[]
n = int(input("Enter total number:"))
for i in range(n):
    L = int(input("Enter values of list:"))
    l.append(L)

print(max_min(l))
```

**2. Write a Python function that takes a positive integer and returns the sum of the
cube of all the positive integers smaller than the specified number**

```python
def cube(n):
    s = 0
    while (n!=0):
        r = n%10
        s = s + r**3
        n = n//10

    return s

n = int(input("Enter number:"))
if n < 0:
    print("Enter positive number!")

else:
  print(f"Sum of cube of all positive number = {cube(n)}")
```

**3. Write a Python function to print 1 to n using recursion. (Note: Do not use loop)**

```python
def call(n):
        if n == 0:
            return 0

        else:
            call(n-1)
            print(n,end = " ")

n = int(input("Enter number:"))
call(n)
```

**4. Write a recursive function to print Fibonacci series upto n terms.**

```python
def fib(n):
    if n <=1:
        return n

    else:
        return fib(n-1) + fib(n-2)

n = int(input("Enter range:"))
for i in range(n):
    print(fib(i), end=" ")
```

**5. Write a lambda function to find volume of cone.**

```python
import math

r = int(input("Enter radius of cone:"))
h = int(input("Enter height of cone:"))

vol = lambda r,h : 1/3 * math.pi * r**2 * h

print(f"Volume of cone:{vol(r,h):.2f}")
```
