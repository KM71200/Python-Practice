**Q1) Scan n values in range 0-3 and print the number of times each value has
occurred.**

**3) WAP to input a list of scores for N students in a list data type. Find the score of
the runner-up and print the output.**

```python
#Name: KARMANYA MALETHA
#Sap Id: 590025695
#Program 3

n = int(input("Enter number of students:"))
L = []

for i in range(n):
    scores = int(input(("Enter scores:")))
    L.append(scores)

L.sort(reverse = True)
print(f"Runner-up Score = {L[1]}")
```

**5) Store details of n movies in a dictionary by taking input from the user. Each
movie must store details like name,  year, director name, production cost,
collection made (earning) & perform the following :-         
a) print all movie details       
b) display name of movies       released before 2015              
c) print movies that made a profit.                           
d) print movies directed by a particular director.**

```python
#Name: KARMANYA MALETHA
#Sap Id: 590025695
#Program 5

movies = {}
n = int(input("Enter number of movies to enter:"))
for i in range(n):
    name = input("Movie name:")
    year = int(input("Enter Year:"))
    d_name = input("Director name:")
    p_cost = float(input("Enter production cost: "))
    earning = float(input("Collection made:"))

    movies[name] = {
        "Year":year,
        "Director":d_name,
        "Production": p_cost,
        "Collection": earning
        }

print(movies)

for k,v in movies.items():
    if v["Year"] < 2015:
        print(f"Movies before 2015:{k}")



for k,v in movies.items():
    if v["Production"] < v["Collection"]:
        print(f"Profit made by:{k}")



n = input("Enter director name:")
for k,v in movies.items():
    if v["Director"] == n:
        print(k)
```
